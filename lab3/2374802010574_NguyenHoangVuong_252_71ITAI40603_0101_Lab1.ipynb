```python
import matplotlib.pyplot as vuong
image_rgb_convert = cv.cvtColor(image, cv.COLOR_BGR2RGB)

vuong.imshow(image_rgb_convert)
vuong.show()
```

```python
import matplotlib.pyplot as vuong
image_rgb_manual = image[:, :, ::-1]

vuong.imshow(image_rgb_manual)
vuong.show()
```

### 5. Crop ảnh (Cắt ảnh)

```python
import matplotlib.pyplot as vuong

height, width, _ = image_rgb_convert.shape
start_row = int(height * 0.25)
end_row = int(height * 0.75)
start_col = int(width * 0.25)
end_col = int(width * 0.75)

cropped_image = image_rgb_convert[start_row:end_row, start_col:end_col]
vuong.imshow(cropped_image)
vuong.title("Ảnh đã Crop")
vuong.show()
```

```python
import cv2 as cv
import matplotlib.pyplot as vuong

zoomed_image_rgb = cv.resize(image_rgb_convert, None, fx=2, fy=2, interpolation=cv.INTER_CUBIC)

h1, w1, _ = image_rgb_convert.shape
h2, w2, _ = zoomed_image_rgb.shape
if h1 != h2:
    resized_zoomed_image = cv.resize(zoomed_image_rgb, (int(w2 * (h1 / h2)), h1), interpolation=cv.INTER_AREA)
else:
    resized_zoomed_image = zoomed_image_rgb
h_concat_image = cv.hconcat([image_rgb_convert, resized_zoomed_image])
print("Ảnh ghép ngang:")
vuong.imshow(h_concat_image)
vuong.title("Ghép ngang 2 ảnh")
vuong.show()

if w1 != w2:
    resized_zoomed_image_v = cv.resize(zoomed_image_rgb, (w1, int(h2 * (w1 / w2))), interpolation=cv.INTER_AREA)
else:
    resized_zoomed_image_v = zoomed_image_rgb

v_concat_image = cv.vconcat([image_rgb_convert, resized_zoomed_image_v])

vuong.imshow(v_concat_image)
vuong.title("Ghép dọc 2 ảnh")
vuong.show()
```
