# LAB 2: THỰC HÀNH IMAGE FILTER

```python
# Khai báo thư viện ảnh
import cv2
import matplotlib.pyplot as Vuong   

# Kiểm tra opencv đọc ảnh và đặt ảnh cho đúng hiển thị ban đầu
img = cv2.imread('cat.jpg')
img = img[:, :, ::-1] 
Vuong.imshow(img)
```

```python
# 1
# áp dụng boxfilter với kernel 10x10 để làm mờ ảnh
blur_box_10 = cv2.blur(img, (10, 10))

Vuong.figure(figsize=(14, 4))
Vuong.subplot(1, 3, 1)
Vuong.imshow(img)
Vuong.title('Original Image')


Vuong.subplot(1, 3, 2)
Vuong.imshow(blur_box_10)
Vuong.title('Box Filter 10x10')
Vuong.show()
```

```python
# 2. Gaussian Filter (kích thước kernel là số lẻ)

gaussian_11 = cv2.GaussianBlur(img, (11, 11), 0)
# Áp dụng Gaussian Filter với kernel 11x11 để làm mờ ảnh

Vuong.figure(figsize=(10, 4))
# Tạo khung hình hiển thị

Vuong.subplot(1, 3, 1)
Vuong.imshow(img)
Vuong.title('Original Image')


Vuong.subplot(1, 3, 2)
Vuong.imshow(gaussian_11)
Vuong.title('Gaussian Filter 11x11')

Vuong.show()

```

```python
# 3. Median Filter
median_5 = cv2.medianBlur(img, 5)
# Áp dụng Median Filter với kernel size = 5

Vuong.figure(figsize=(10, 4))

Vuong.subplot(1, 3, 1)
Vuong.imshow(img)
Vuong.title('Original Image')


Vuong.subplot(1, 3, 2)
Vuong.imshow(median_5)
Vuong.title('Median Filter 5x5')

Vuong.show()
# Hiển thị hình ảnh
```

```python
# 4. Áp dụng 2 bộ lọc (Box Filter và Gaussian Filter)

# Đọc ảnh
lenna = cv2.imread("lenna.jpg")
tiger = cv2.imread("tiger.jpg")
# Chuyển BGR -> RGB
lenna = cv2.cvtColor(lenna, cv2.COLOR_BGR2RGB)
tiger = cv2.cvtColor(tiger, cv2.COLOR_BGR2RGB)

# Box Filter
lenna_box = cv2.blur(lenna, (7, 7))
tiger_box = cv2.blur(tiger, (7, 7))
# Gaussian Filter
lenna_gaussian = cv2.GaussianBlur(lenna, (7, 7), 0)
tiger_gaussian = cv2.GaussianBlur(tiger, (7, 7), 0)

# Hiển thị kết quả
Vuong.figure(figsize=(10, 6))

Vuong.subplot(2, 2, 1)
Vuong.imshow(lenna_box)
Vuong.title("Lenna - Box Filter")

Vuong.subplot(2, 2, 2)
Vuong.imshow(lenna_gaussian)
Vuong.title("Lenna - Gaussian Filter")

Vuong.subplot(2, 2, 3)
Vuong.imshow(tiger_box)
Vuong.title("Tiger - Box Filter")

Vuong.subplot(2, 2, 4)
Vuong.imshow(tiger_gaussian)
Vuong.title("Tiger - Gaussian Filter")

Vuong.show()
```

### 5. Nhận xét về các kích thước filter
#### Box Filter

- Làm mở ảnh bằng cách lấy trung bình các pixel lân cận, giúp giảm nhiễu nhanh nhưng làm mất nhiều chi tiết và biên ảnh.

#### Gaussian Filter

- Sử dụng trọng số theo phân phối Gaussian nên cho ảnh mượt hơn và giữ biên tốt hơn so với Box Filter.

KẾT LUẬN: Trong hai bộ lọc đã sử dụng, Gaussian Filter cho chất lượng ảnh tốt hơn và phù hợp hơn trong hầu hết các trường hợp xử lý ảnh cơ bản.

```python
# 6. Histogram Equalization
img = cv2.imread("tiger.jpg")

# Áp dụng Median Blur để làm mờ ảnh
blur = cv2.medianBlur(img,5)

# Chuyển sang ảnh xám
blur = cv2.cvtColor(blur, cv2.COLOR_BGR2GRAY)

# Cân bằng histogram
equalized = cv2.equalizeHist(blur)

Vuong.figure(figsize=(8, 4))

Vuong.subplot(1, 2, 1)
Vuong.imshow(blur, cmap='gray')
Vuong.title("Original Grayscale")

Vuong.subplot(1, 2, 2)
Vuong.imshow(equalized, cmap='gray')
Vuong.title("Histogram Equalized")

Vuong.show()
```

```python
import cv2
import numpy as np


# Đọc ảnh
img1 = cv2.imread('exT3_01.png')
img2 = cv2.imread('exT3_02.png')

# Cắt ảnh
img1 = img1[:155, 1:191]
img2 = img2[:155, :190]

# Grayscale
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

# Lấy hiệu
diff = cv2.absdiff(gray1, gray2)

# Threshold
_, thresh = cv2.threshold(diff, 30, 255, cv2.THRESH_BINARY)

# Morphology
kernel = np.ones((5,5), np.uint8)
thresh = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel)
thresh = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel)

# Mask lấy quả bóng
result = cv2.bitwise_and(img1, img1, mask=thresh)

# Hiển thị
cv2.imshow("Difference", diff)
cv2.imshow("Mask qua bong", thresh)
cv2.imshow("Ket qua chi con qua bong", result)

cv2.waitKey(0)
cv2.destroyAllWindows()

```

## BAI TAP MO RONG

```python
img_bgr = cv2.imread('cat.jpg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Display the image to confirm successful loading and conversion
Vuong.figure(figsize=(6, 6))
Vuong.imshow(img_rgb)
Vuong.title('Loaded Image (RGB)')
Vuong.axis('off')
Vuong.show()
```

```python
gray_img = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

backward_diff_x = np.zeros_like(gray_img, dtype=np.float32)
backward_diff_x[:, 1:] = gray_img[:, 1:].astype(np.float32) - gray_img[:, :-1].astype(np.float32)

backward_diff_y = np.zeros_like(gray_img, dtype=np.float32)
backward_diff_y[1:, :] = gray_img[1:, :].astype(np.float32) - gray_img[:-1, :].astype(np.float32)

forward_diff_x = np.zeros_like(gray_img, dtype=np.float32)
forward_diff_x[:, :-1] = gray_img[:, 1:].astype(np.float32) - gray_img[:, :-1].astype(np.float32)

forward_diff_y = np.zeros_like(gray_img, dtype=np.float32)
forward_diff_y[:-1, :] = gray_img[1:, :].astype(np.float32) - gray_img[:-1, :].astype(np.float32)

central_diff_x = np.zeros_like(gray_img, dtype=np.float32)
central_diff_x[:, 1:-1] = (gray_img[:, 2:].astype(np.float32) - gray_img[:, :-2].astype(np.float32)) / 2.0

central_diff_y = np.zeros_like(gray_img, dtype=np.float32)
central_diff_y[1:-1, :] = (gray_img[2:, :].astype(np.float32) - gray_img[:-2, :].astype(np.float32)) / 2.0

backward_gradient_magnitude = np.sqrt(backward_diff_x**2 + backward_diff_y**2)
forward_gradient_magnitude = np.sqrt(forward_diff_x**2 + forward_diff_y**2)
central_gradient_magnitude = np.sqrt(central_diff_x**2 + central_diff_y**2)

Vuong.figure(figsize=(18, 5))

Vuong.subplot(1, 4, 1)
Vuong.imshow(gray_img, cmap='gray')
Vuong.title('Grayscale Image')
Vuong.axis('off')

Vuong.subplot(1, 4, 2)
Vuong.imshow(backward_gradient_magnitude, cmap='gray')
Vuong.title('Backward Gradient Magnitude')
Vuong.axis('off')

Vuong.subplot(1, 4, 3)
Vuong.imshow(forward_gradient_magnitude, cmap='gray')
Vuong.title('Forward Gradient Magnitude')
Vuong.axis('off')

Vuong.subplot(1, 4, 4)
Vuong.imshow(central_gradient_magnitude, cmap='gray')
Vuong.title('Central Gradient Magnitude')
Vuong.axis('off')

Vuong.tight_layout()
Vuong.show()
```

```python
import numpy as np
import cv2


finite_diff_x_component = np.zeros_like(gray_img, dtype=np.float32)
finite_diff_x_component[:, 1:] = np.abs(gray_img[:, 1:].astype(np.float32) - gray_img[:, :-1].astype(np.float32))

finite_diff_y_component = np.zeros_like(gray_img, dtype=np.float32)
finite_diff_y_component[1:, :] = np.abs(gray_img[1:, :].astype(np.float32) - gray_img[:-1, :].astype(np.float32))

finite_difference_image = np.sqrt(finite_diff_x_component**2 + finite_diff_y_component**2)

Vuong.figure(figsize=(12, 6))

Vuong.subplot(1, 2, 1)
Vuong.imshow(gray_img, cmap='gray')
Vuong.title('Original Grayscale Image')
Vuong.axis('off')

Vuong.subplot(1, 2, 2)
Vuong.imshow(finite_difference_image, cmap='gray')
Vuong.title('Finite Difference Filter (Magnitude)')
Vuong.axis('off')

Vuong.tight_layout()
Vuong.show()

print("Finite Difference Filter applied and displayed.")
```

```python
gaussian_blurred_3x3 = cv2.GaussianBlur(img_rgb, (3, 3), 0)
gaussian_blurred_5x5 = cv2.GaussianBlur(img_rgb, (5, 5), 0)
gaussian_blurred_9x9 = cv2.GaussianBlur(img_rgb, (9, 9), 0)

Vuong.figure(figsize=(15, 5))

Vuong.subplot(1, 4, 1)
Vuong.imshow(img_rgb)
Vuong.title('Original Image')
Vuong.axis('off')

Vuong.subplot(1, 4, 2)
Vuong.imshow(gaussian_blurred_3x3)
Vuong.title('Gaussian Blurred (3x3)')
Vuong.axis('off')

Vuong.subplot(1, 4, 3)
Vuong.imshow(gaussian_blurred_5x5)
Vuong.title('Gaussian Blurred (5x5)')
Vuong.axis('off')

Vuong.subplot(1, 4, 4)
Vuong.imshow(gaussian_blurred_9x9)
Vuong.title('Gaussian Blurred (9x9)')
Vuong.axis('off')

Vuong.tight_layout()
Vuong.show()
```

```python
gray_img_sobel = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

# Apply Sobel operator in x-direction
sobelx = cv2.Sobel(gray_img_sobel, cv2.CV_64F, 1, 0, ksize=3)

# Apply Sobel operator in y-direction
sobely = cv2.Sobel(gray_img_sobel, cv2.CV_64F, 0, 1, ksize=3)

# Convert absolute values to 8-bit unsigned integers
abs_sobelx = cv2.convertScaleAbs(sobelx)
abs_sobely = cv2.convertScaleAbs(sobely)

# Combine the x and y gradients
sobel_combined = cv2.addWeighted(abs_sobelx, 0.5, abs_sobely, 0.5, 0)

# Display the results
Vuong.figure(figsize=(15, 8))

Vuong.subplot(2, 2, 1)
Vuong.imshow(gray_img_sobel, cmap='gray')
Vuong.title('Grayscale Image')
Vuong.axis('off')

Vuong.subplot(2, 2, 2)
Vuong.imshow(abs_sobelx, cmap='gray')
Vuong.title('Sobel X')
Vuong.axis('off')

Vuong.subplot(2, 2, 3)
Vuong.imshow(abs_sobely, cmap='gray')
Vuong.title('Sobel Y')
Vuong.axis('off')

Vuong.subplot(2, 2, 4)
Vuong.imshow(sobel_combined, cmap='gray')
Vuong.title('Sobel Combined')
Vuong.axis('off')

Vuong.tight_layout()
Vuong.show()
```

```python
gray_img_canny = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2GRAY)

edges_low_thresh = cv2.Canny(gray_img_canny, 50, 150)
edges_mid_thresh = cv2.Canny(gray_img_canny, 100, 200)
edges_high_thresh = cv2.Canny(gray_img_canny, 150, 250)

Vuong.figure(figsize=(15, 5))

Vuong.subplot(1, 4, 1)
Vuong.imshow(gray_img_canny, cmap='gray')
Vuong.title('Grayscale Image')
Vuong.axis('off')

Vuong.subplot(1, 4, 2)
Vuong.imshow(edges_low_thresh, cmap='gray')
Vuong.title('Canny (Low Thresh)')
Vuong.axis('off')

Vuong.subplot(1, 4, 3)
Vuong.imshow(edges_mid_thresh, cmap='gray')
Vuong.title('Canny (Mid Thresh)')
Vuong.axis('off')

Vuong.subplot(1, 4, 4)
Vuong.imshow(edges_high_thresh, cmap='gray')
Vuong.title('Canny (High Thresh)')
Vuong.axis('off')

Vuong.tight_layout()
Vuong.show()
```

```python
import numpy as np

edges_for_hough = edges_mid_thresh

lines = cv2.HoughLinesP(edges_for_hough, 1, np.pi / 180, 100, minLineLength=100, maxLineGap=10)

line_image = np.copy(img_rgb) * 0

if lines is not None:
    for line in lines:
        x1, y1, x2, y2 = line[0]
        cv2.line(line_image, (x1, y1), (x2, y2), (255, 0, 0), 5)

lines_on_original = cv2.addWeighted(img_rgb, 0.8, line_image, 1, 0)

Vuong.figure(figsize=(12, 6))

Vuong.subplot(1, 3, 1)
Vuong.imshow(img_rgb)
Vuong.title('Original Image')
Vuong.axis('off')

Vuong.subplot(1, 3, 2)
Vuong.imshow(edges_for_hough, cmap='gray')
Vuong.title('Canny Edges for Hough')
Vuong.axis('off')

Vuong.subplot(1, 3, 3)
Vuong.imshow(lines_on_original)
Vuong.title('Hough Lines Detected')
Vuong.axis('off')

Vuong.tight_layout()
Vuong.show()
```
